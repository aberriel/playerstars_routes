```python
from playerstars_domain import Player
from playerstars_adapters import PlayerAdapter

adapter = PlayerAdapter(table_name='player_dev')
players = adapter.filter(premium_expiration_date__exists=True)

from boto3.dynamodb.conditions import Attr
raw_players = adapter._table.scan(FilterExpression=Attr('premium_expiration_date').exists())['Items']

for raw_player in raw_players:
    del raw_player['premium_expiration_date']
    player = Player.from_json(raw_player)
    player.set_adapter(adapter)
    player.save()
```
